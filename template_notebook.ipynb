```python
def sr_3d_visualization(sr_pos, box_size=100.0, downsample_factor=64^3):
    sample_size = int(1/downsample_factor * 134217728)
    if sample_size > 0:
        sr_pos = sr_pos[np.random.choice(134217728, sample_size, replace=False)]
    x_sr, y_sr, z_sr = sr_pos[:,0], sr_pos[:,1], sr_pos[:,2]
    fig = plt.figure(figsize=(12, 18))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(x_sr, y_sr, z_sr, alpha=0.5, s=1)
    ax.set_xlabel('X Position [Mpc/h]')
    ax.set_ylabel('Y Position [Mpc/h]')
    ax.set_zlabel('Z Position [Mpc/h]')
    
    plt.show()
```

```python
def plot_3d_positions(positions, sample_size=64^3):
    # If positions is (N, N, N, 3), reshape to (N³, 3)
    if len(positions.shape) == 4:
        positions = positions.reshape(-1, 3)
    
    # Now positions has shape (N³, 3) where:
    # - Each row represents one particle
    # - Column 0 is x-coordinate
    # - Column 1 is y-coordinate
    # - Column 2 is z-coordinate
    
    # Random sampling for visualization
    total_points = 512^3
    indices = np.random.choice(positions.shape[0], min(sample_size, total_points), replace=False)
    plot_points = positions[indices]  # Shape: (sample_size, 3)
    
    fig = plt.figure(figsize=(10, 10))
    ax = fig.add_subplot(111, projection='3d')
    
    ax.scatter(
        plot_points[:, 0],  # All x coordinates
        plot_points[:, 1],  # All y coordinates
        plot_points[:, 2],  # All z coordinates
        alpha=0.6,
        s=1
    )
```
